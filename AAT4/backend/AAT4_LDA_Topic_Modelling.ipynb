# AAT4 — Topic Modelling with LDA
Fit LDA on a tiny corpus and inspect the top words.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
docs = ['Neural networks learn from training data.', 'Deep learning uses layers and weights.', 'Rainfall and storms affect climate.', 'Weather models use satellite observations.', 'Search engines retrieve relevant documents.']
vectorizer = CountVectorizer(stop_words='english')
X = vectorizer.fit_transform(docs)
lda = LatentDirichletAllocation(n_components=2, random_state=21).fit(X)
words = vectorizer.get_feature_names_out()
for i, weights in enumerate(lda.components_):
    top = weights.argsort()[-5:][::-1]
    print('Topic', i + 1, [words[j] for j in top])
```
