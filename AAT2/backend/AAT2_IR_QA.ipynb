# AAT2 — Information Retrieval and Question Answering
Rank local documents using TF-IDF and cosine similarity.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
docs = ['NLP processes human language.', 'TF-IDF weights informative words.', 'NER finds names and locations.']
question = 'What does TF-IDF do?'
vectorizer = TfidfVectorizer(stop_words='english')
matrix = vectorizer.fit_transform(docs + [question])
scores = cosine_similarity(matrix[-1], matrix[:-1]).ravel()
sorted(zip(scores, docs), reverse=True)
```
