# AAT5 — Word-Sense Disambiguation with Lesk
Download WordNet resources before running. Lesk is a heuristic, not a guaranteed answer.

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.wsd import lesk
sentence = 'She deposited her salary at the bank near the river.'
tokens = nltk.word_tokenize(sentence)
sense = lesk(tokens, 'bank')
print(sense.name() if sense else 'No sense found')
print(sense.definition() if sense else 'Try adding context')
```
