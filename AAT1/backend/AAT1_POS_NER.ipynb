# AAT1 — POS Tagging and Named Entity Recognition
Install spaCy and its small English model, then run the cells.

```python
import spacy
nlp = spacy.load('en_core_web_sm')
text = 'Dr. Asha Rao joined Google in Bengaluru on Monday.'
doc = nlp(text)
[(token.text, token.pos_, token.tag_, token.lemma_) for token in doc]
```

```python
[(ent.text, ent.label_, spacy.explain(ent.label_)) for ent in doc.ents]
```
