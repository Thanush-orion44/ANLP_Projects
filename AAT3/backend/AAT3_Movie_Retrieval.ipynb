# AAT3 — Movie Knowledge Retrieval
This educational RAG-style baseline retrieves local records and uses the records as evidence; no generative model is called.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
movies = [{'title':'Orbit Nine','plot':'A crew at a space research station investigates a signal.'}, {'title':'Paper Kingdom','plot':'An archivist finds coded messages in old books.'}]
docs = [m['title'] + ' ' + m['plot'] for m in movies]
question = 'Which story is about a space station?'
v = TfidfVectorizer(stop_words='english')
x = v.fit_transform(docs + [question])
scores = cosine_similarity(x[-1], x[:-1]).ravel()
[(movies[i]['title'], round(float(scores[i]), 3), movies[i]['plot']) for i in scores.argsort()[::-1]]
```
